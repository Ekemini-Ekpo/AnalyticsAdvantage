```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load datasets
# Update paths if your files are in a subfolder, e.g. '/content/drive/MyDrive/AnalyticsAdvantage/'
bigmart = pd.read_csv('/content/drive/MyDrive/Big Mart Dataset.csv')
marketing = pd.read_csv('/content/drive/MyDrive/MarketingCampaignEfficiency.csv')

print('Big Mart shape:', bigmart.shape)
print('Marketing shape:', marketing.shape)
bigmart.head()
```

## Step 2: Basic Histogram with Matplotlib

```python
# Histogram of Item MRP (price)
plt.figure(figsize=(8, 5))
plt.hist(bigmart['Item_MRP'], bins=30, color='steelblue', edgecolor='white')
plt.title('Distribution of Item MRP')
plt.xlabel('Item MRP')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

## Step 3: Histogram of Sales

```python
# Histogram of Item Outlet Sales
plt.figure(figsize=(8, 5))
plt.hist(bigmart['Item_Outlet_Sales'], bins=40, color='coral', edgecolor='white')
plt.title('Distribution of Item Outlet Sales')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

## Step 4: Histogram with Seaborn (KDE overlay)

```python
# Seaborn histogram with KDE curve
plt.figure(figsize=(8, 5))
sns.histplot(bigmart['Item_MRP'], bins=30, kde=True, color='steelblue')
plt.title('Item MRP — Histogram with KDE')
plt.xlabel('Item MRP')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

## Step 5: Marketing Campaign — Sales by Promotion

```python
# Histogram of SalesInThousands for each promotion
fig, axes = plt.subplots(1, 3, figsize=(14, 5), sharey=True)

for i, promo in enumerate(sorted(marketing['Promotion'].unique())):
    subset = marketing[marketing['Promotion'] == promo]['SalesInThousands']
    axes[i].hist(subset, bins=20, color=['steelblue', 'coral', 'seagreen'][i], edgecolor='white')
    axes[i].set_title(f'Promotion {promo}')
    axes[i].set_xlabel('Sales (thousands)')
    if i == 0:
        axes[i].set_ylabel('Frequency')

fig.suptitle('Sales Distribution by Promotion Type', fontsize=14)
plt.tight_layout()
plt.show()
```

## Step 6: Overlapping Histograms for Comparison

```python
# Overlapping histograms — all promotions on one chart
plt.figure(figsize=(9, 5))
colors = ['steelblue', 'coral', 'seagreen']

for promo, color in zip(sorted(marketing['Promotion'].unique()), colors):
    subset = marketing[marketing['Promotion'] == promo]['SalesInThousands']
    plt.hist(subset, bins=20, alpha=0.5, label=f'Promotion {promo}', color=color, edgecolor='white')

plt.title('Overlapping Sales Distributions by Promotion')
plt.xlabel('Sales (thousands)')
plt.ylabel('Frequency')
plt.legend()
plt.tight_layout()
plt.show()
```
